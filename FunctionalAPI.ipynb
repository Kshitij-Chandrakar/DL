```python
from keras.models import Model #import Functional API
```

```python
model =Model (inputs =x,outputs=[output1,output2]) #here we will define the input and output layers of the model
```

```python
from keras.layers import* 
```

```python
#Architecture design of the model 
x=Input(shape=(3,))
```

```python
hidden1=Dense(128,activation='relu')(x) #x is the input layer and output1 is the first hidden layer
hidden2=Dense(64,activation='relu')(hidden1) #hidden1 is the first hidden layer and hidden2 is the second hidden layer
```

```python
#now branching the model
output1=Dense(1,activation='linear')(hidden2) #this is the first branch of the model
output2=Dense(1,activation='sigmoid')(hidden2) #this is the second branch of the model
```
