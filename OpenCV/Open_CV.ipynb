```python
import matplotlib.pyplot as plt
import cv2
```

```python
image =cv2.imread("D:\cat.jpg")
#to read the image 
```

```python
plt.imshow(image) #Prepare the image for display
plt.title("Cat")    #to give the title to the image
plt.axis("off") #to hide the axis
plt.show() #to display the image
```

```python
#As we can see the image is not in the correct color format. This is because OpenCV reads the image in BGR format, while Matplotlib expects it in RGB format. To fix this, we can convert the image from BGR to RGB using the cv2.cvtColor() function.  
img=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
plt.imshow(img)
```

```python
cv2.imwrite("cat1.jpg",img) #to save the image in the current directory.
```

```python
image.shape #to get the dimensions of the image. It returns a tuple of (height, width, channels).
```

```python
gray=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY) #to convert the image to grayscale
plt.imshow(gray ,cmap="gray") #to display the grayscale image
```

```python
cv2.imwrite("catgray.png",gray)
```

  ### image resizing

### Cropping

```python
cropped=image[50:200,100:300] #to crop the image. The first two values are the y-coordinates and the last two values are the x-coordinates.
plt.imshow(cropped)
plt.axis("off")
```
