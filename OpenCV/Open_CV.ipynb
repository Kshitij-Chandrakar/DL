```python
import matplotlib.pyplot as plt
import cv2
```

```python
image =cv2.imread("D:\cat.jpg")
#to read the image 
```

```python
plt.imshow(image) #Prepare the image for display
plt.title("Cat")    #to give the title to the image
plt.axis("off") #to hide the axis
plt.show() #to display the image
```

```python
#As we can see the image is not in the correct color format. This is because OpenCV reads the image in BGR format, while Matplotlib expects it in RGB format. To fix this, we can convert the image from BGR to RGB using the cv2.cvtColor() function.  
img=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
plt.imshow(img)
```

```python
cv2.imwrite("cat1.jpg",img) #to save the image in the current directory.
```

```python
image.shape #to get the dimensions of the image. It returns a tuple of (height, width, channels).
```

```python
gray=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY) #to convert the image to grayscale
plt.imshow(gray ,cmap="gray") #to display the grayscale image
```

```python
cv2.imwrite("catgray.png",gray)
```

## image Transformation 

  ### image resizing 

```python
resize=cv2.resize(image,(500,200))
plt.imshow(resize)
```

### Cropping

```python
cropped=image[50:200,100:300] #to crop the image. The first two values are the y-coordinates and the last two values are the x-coordinates.
plt.imshow(cropped)
plt.axis("off")
```

### Rotation

```python
image.shape
```

```python
centre=(image.shape[1]//2,image.shape[0]//2) #to get the center of the image. The first value is the x-coordinate and the second value is the y-coordinate.
M=cv2.getRotationMatrix2D(centre,45,1) #to get the rotation matrix. The first value is the center of rotation, the second value is the angle of rotation in degrees, and the third value is the scale factor.
rotate=cv2.warpAffine(image,M,(image.shape[1],image.shape[0]))
```

```python
plt.imshow(rotate)
plt.axis("off")
```

### Flipping

```python
flipped_h=cv2.flip(image,1)
plt.imshow(flipped_h)
plt.axis("off")
```

```python
flipped_h=cv2.flip(image,0)
plt.imshow(flipped_h)
plt.axis("off")
```

```python
flipped_h=cv2.flip(image,-1)
plt.imshow(flipped_h)
plt.axis("off")
```

### Drawing a line

```python
pt1=(50,100)
pt2=(300,150)
cv2.line(image,pt1,pt2,(0,255,0),5) #to draw a line on the image. The first value is the image, the second value is the starting point of the line, the third value is the ending point of the line, the fourth value is the color of the line in BGR format, and the fifth value is the thickness of the line.
plt.imshow(image)
```

```python
cv2.rectangle(img,(50,50),(200,200),(255,0,0),5) #to draw a rectangle on the image. The first value is the image, the second value is the top-left corner of the rectangle, the third value is the bottom-right corner of the rectangle, the fourth value is the color of the rectangle in BGR format, and the fifth value is the thickness of the rectangle.
plt.imshow(img)
```

```python
##to add text 
font=cv2.FONT_HERSHEY_SIMPLEX
cv2.putText(gray,"Cat",(90,90),font,1,(255,0,0),6) #to add text to the image. The first value is the image, the second value is the text to be added, the third value is the bottom-left corner of the text, the fourth value is the font, the fifth value is the font scale, the sixth value is the color of the text in BGR format, and the seventh value is the thickness of the text.
plt.imshow(gray,cmap="gray")
```
